```python

n = int(input("Enter number of items: "))

profit = []
weight = []

for i in range(n):
    p = int(input("Enter profit: "))
    w = int(input("Enter weight: "))

    profit.append(p)
    weight.append(w)

capacity = int(input("Enter knapsack capacity: "))

ratio = []

for i in range(n):
    ratio.append(profit[i] / weight[i])

# Sort items by profit/weight ratio
items = sorted(
    range(n),
    key=lambda i: ratio[i],
    reverse=True
)

total_profit = 0

for i in items:
    if capacity == 0:
        break

    if weight[i] <= capacity:
        total_profit += profit[i]
        capacity -= weight[i]

    else:
        total_profit += ratio[i] * capacity
        capacity = 0

print("Maximum Profit:", total_profit)
```

# **A(3) Fractional Knapsack Problem Using Greedy Method**

## **1. Concept of Fractional Knapsack**

**Fractional Knapsack is a greedy algorithm used to maximize the total profit by selecting items within a given weight capacity of a knapsack.**

The important point is that **we can take a fraction (part) of an item** instead of taking the complete item.

### **1. Understand with a real-life example**

Suppose you have a bag with a maximum capacity of **50 kg** and three items.

| Item | Weight | Profit |
| ---- | ------ | ------ |
| A    | 10 kg  | ₹60    |
| B    | 20 kg  | ₹100   |
| C    | 30 kg  | ₹120   |

Your objective is to put items in the bag and earn maximum profit without exceeding 50 kg.

### **2. What is the Greedy Strategy?**

We calculate the profit per unit weight for every item.

Formula:

\\[ \text{Profit/Weight Ratio}=\frac{\text{Profit}}{\text{Weight}} \\]

| Item | Weight | Profit | Ratio |
| ---- | ------ | ------ | ----- |
| A    | 10     | 60     | 6     |
| B    | 20     | 100    | 5     |
| C    | 30     | 120    | 4     |

Now sort items in descending order of their profit/weight ratio.

So the priority becomes:

1. A → Ratio 6
2. B → Ratio 5
3. C → Ratio 4

**Greedy rule:** Always select the item with the highest profit per unit weight first.

### **3. Step-by-step solution**

**Knapsack capacity = 50 kg**

1. **Select item A**
   - Weight = 10 kg
   - Profit = ₹60
   - Remaining capacity = 40 kg
   - Total profit = ₹60
2. **Select item B**
   - Weight = 20 kg
   - Profit = ₹100
   - Remaining capacity = 20 kg
   - Total profit = ₹160
3. **Select item C partially**

   Only 20 kg capacity is remaining, but C weighs 30 kg.

   Fraction selected:

   \\[ \frac{20}{30}=\frac{2}{3} \\]

   Profit earned:

   \\[ 120\times\frac{20}{30}=80 \\]

   Total profit = ₹160 + ₹80 = ₹240.

**Maximum profit**

# **₹240**

Total weight = 10 + 20 + 20 = 50 kg.

### **4. Difference between Fractional and 0/1 Knapsack**

| Fractional Knapsack                     | 0/1 Knapsack                                         |
| --------------------------------------- | ---------------------------------------------------- |
| Items can be divided.                   | Items cannot be divided.                             |
| Can take partial items.                 | Take entire item or reject it.                       |
| Greedy method gives optimal solution.   | Greedy method does not always give optimal solution. |
| Generally solved using greedy strategy. | Commonly solved using Dynamic Programming.           |

## **5. Algorithm**

1. Start.
2. Accept the number of items.
3. Accept weight and profit of each item.
4. Calculate the profit/weight ratio.
5. Sort items in descending order of ratio.
6. Select items having the highest ratio first.
7. If the complete item fits, take it.
8. Otherwise, take the fraction that fits.
9. Calculate the maximum profit.
10. Display the result and stop.



## **8. Complexity Analysis**

Let \\(n\\) be the number of items.

| Operation          | Time Complexity  |
| ------------------ | ---------------- |
| Taking input       | \\(O(n)\\)       |
| Calculating ratios | \\(O(n)\\)       |
| Sorting items      | \\(O(n\log n)\\) |
| Selecting items    | \\(O(n)\\)       |
| **Overall**        | \\(O(n\log n)\\) |

**Space Complexity:** \\(O(n)\\) because we use arrays/lists to store weights, profits, ratios and sorted item indices.

## **9. Conclusion**

The Fractional Knapsack problem is solved using a greedy strategy. We select items in descending order of their profit-to-weight ratio. If an item cannot fit completely, we take only the required fraction. This approach maximizes profit and has a time complexity of \\(O(n\log n)\\).
