````python

def is_safe(board, row, col, n):

    # Check upper column
    for i in range(row):
        if board[i][col] == 1:
            return False

    # Check upper left diagonal
    i = row - 1
    j = col - 1

    while i >= 0 and j >= 0:
        if board[i][j] == 1:
            return False
        i -= 1
        j -= 1

    # Check upper right diagonal
    i = row - 1
    j = col + 1

    while i >= 0 and j < n:
        if board[i][j] == 1:
            return False
        i -= 1
        j += 1

    return True


def solve_n_queens(board, row, n):

    if row == n:
        return True

    for col in range(n):

        if is_safe(board, row, col, n):

            board[row][col] = 1

            if solve_n_queens(board, row + 1, n):
                return True

            # Backtracking
            board[row][col] = 0

    return False


n = int(input("Enter number of queens: "))

board = [[0 for j in range(n)] for i in range(n)]

if solve_n_queens(board, 0, n):

    print("N-Queens Solution:")

    for row in board:
        print(row)

else:
    print("Solution does not exist")
````

# **A(5) N-Queens Problem Using Backtracking**

## **1. Concept of N-Queens Problem**

**The N-Queens problem is a classic problem in which we have to place N queens on an N × N chessboard such that no two queens attack each other.**

We use the **Backtracking algorithm** to solve this problem.

### **2. Rules of N-Queens**

In chess, a Queen can attack another Queen in three directions:

1. Same row (horizontal).
2. Same column (vertical).
3. Same diagonal.

Therefore, we must place queens so that:

- No two queens are in the same row.
- No two queens are in the same column.
- No two queens are on the same diagonal.

### **3. What is Backtracking?**

**Backtracking is an algorithmic technique that builds a solution step by step. If a choice leads to an invalid solution, it goes back and tries another choice.**

In N-Queens:

1. Start with the first row.
2. Place a Queen in a safe column.
3. Move to the next row.
4. Check whether the next Queen can be placed safely.
5. If no safe position exists, remove the previous Queen.
6. Try another column.
7. Continue until all N Queens are placed.

This process is called Backtracking.

## **8. Complexity Analysis**

| Complexity       | Value        |
| ---------------- | ------------ |
| Time Complexity  | \\(O(N!)\\)  |
| Space Complexity | \\(O(N^2)\\) |

- **Time:** In the worst case, the algorithm explores many possible Queen arrangements.
- **Space:** The chessboard requires \\(O(N^2)\\) space, while recursion requires \\(O(N)\\) additional stack space.

## **9. Conclusion**

The N-Queens problem is solved using the Backtracking technique. Queens are placed row by row after checking whether each position is safe. If a position leads to an invalid solution, the algorithm removes the Queen and tries another position. The process continues until all Queens are placed successfully.
