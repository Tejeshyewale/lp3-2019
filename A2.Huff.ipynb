````python

import heapq

def huffman_encoding(data):
    frequency = {}

    for char in data:
        frequency[char] = frequency.get(char, 0) + 1

    heap = [[freq, [[char, ""]]] for char, freq in frequency.items()]
    heapq.heapify(heap)

    if len(heap) == 1:
        return {heap[0][1][0][0]: "0"}

    while len(heap) > 1:
        left = heapq.heappop(heap)
        right = heapq.heappop(heap)

        left_codes = [[char, "0" + code] for char, code in left[1]]
        right_codes = [[char, "1" + code] for char, code in right[1]]

        merged = left_codes + right_codes

        heapq.heappush(heap, [left[0] + right[0], merged])

    return {char: code for char, code in heap[0][1]}


text = input("Enter a string: ")

codes = huffman_encoding(text)

print("\nHuffman Codes:")

for char, code in sorted(codes.items()):
    print(char, ":", code)
````

# **Huffman Encoding – Concept Explained from Basics**

Let's understand Huffman Encoding with a real-life example before looking at the code.

## **1. What is Huffman Encoding?**

**Huffman Encoding is a lossless data compression technique that reduces the number of bits required to store data.**

It assigns:

- **Short binary codes** to frequently occurring characters.
- **Long binary codes** to less frequently occurring characters.

The main objective is to save storage space without losing any original information.

## **2. Why do we need Huffman Encoding?**

Suppose you want to store this text:

```
AAAAAABBBCCD
```

Normally, computers use 8 bits (1 byte) for each ASCII character.

There are 12 characters.

Therefore:

\\[ 12 \times 8 = 96 \text{ bits} \\]

But what if we use shorter binary codes for frequently occurring characters?

We can reduce the total number of bits required.

## **3. Understand frequency of characters**

First, count how many times each character appears.

Input:

```
AAAAAABBBCCD
```

Frequency table:

| Character | Frequency |
| --------- | --------- |
| A         | 6         |
| B         | 3         |
| C         | 2         |
| D         | 1         |

Notice:

- A occurs 6 times (most frequent).
- D occurs only once (least frequent).

So, A should get a shorter code than D.

## **4. Assign binary codes**

Suppose we assign the following Huffman codes:

This is the basic idea behind Huffman Encoding.

## **5. How does the Greedy Algorithm work?**

This is the most important part for your practical examination.

Huffman Encoding uses a **Greedy Algorithm**.

Greedy means making the best available choice at every step.

In Huffman Encoding, we repeatedly select the two characters/nodes with the smallest frequencies.

Let's take this example:

| Character | Frequency |
| --------- | --------- |
| A         | 5         |
| B         | 9         |
| C         | 12        |
| D         | 13        |
| E         | 16        |
| F         | 45        |

**Step 1:** Select the two smallest frequencies.

A = 5 and B = 9.

Combine them:

\\[ 5+9=14 \\]

**Step 2:** Now the frequencies are:

12, 13, 14, 16, 45.

Select 12 and 13.

\\[ 12+13=25 \\]

**Step 3:** Now frequencies are:

14, 16, 25, 45.

Select 14 and 16.

\\[ 14+16=30 \\]

**Step 4:** Select 25 and 30.

\\[ 25+30=55 \\]

**Step 5:** Select 45 and 55.

\\[ 45+55=100 \\]

Now only one node remains. The Huffman Tree is complete.

Rules for assigning codes:

- Left branch = 0
- Right branch = 1
- Start from the root and follow the path to each character.
- The path gives that character's binary code.

For example, if a character's path is Left → Right → Left, its code is `010`.

## **7. Important properties**

| Property                 | Explanation                  |
| ------------------------ | ---------------------------- |
| Algorithm                | Greedy                       |
| Technique                | Lossless compression         |
| Data structure           | Min Heap                     |
| Output                   | Variable-length binary codes |
| Most frequent character  | Generally shortest code      |
| Least frequent character | Generally longest code       |
| Coding type              | Prefix-free                  |

**What is Prefix-Free?**

No character's Huffman code is the prefix (starting part) of another character's code.

For example:

- A = `0`
- B = `10`
- C = `110`
- D = `111`

Here, no code is the prefix of another code. Therefore, decoding is unambiguous.

## **8. Time and Space Complexity**

Let:

- \\(n\\) = Total number of characters.
- \\(k\\) = Number of unique characters.

| Operation             | Complexity                  |
| --------------------- | --------------------------- |
| Frequency calculation | \\(O(n)\\)                  |
| Building Min Heap     | \\(O(k)\\)                  |
| Combining nodes       | \\(O(k\log k)\\)            |
| Overall time          | \\(O(n+k\log k)\\)          |
| Space                 | \\(O(k)\\), excluding input |
