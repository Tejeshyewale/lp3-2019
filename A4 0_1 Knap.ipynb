```python

n = int(input("Enter number of items: "))

profit = []
weight = []

for i in range(n):
    p = int(input("Enter profit: "))
    w = int(input("Enter weight: "))

    profit.append(p)
    weight.append(w)

capacity = int(input("Enter knapsack capacity: "))

# Create DP table
dp = [[0 for j in range(capacity + 1)]
      for i in range(n + 1)]

# Fill DP table
for i in range(1, n + 1):
    for w in range(1, capacity + 1):

        if weight[i - 1] <= w:
            dp[i][w] = max(
                profit[i - 1] + dp[i - 1][w - weight[i - 1]],
                dp[i - 1][w]
            )

        else:
            dp[i][w] = dp[i - 1][w]

print("Maximum Profit:", dp[n][capacity])
```

# **A(4) 0/1 Knapsack Problem Using Dynamic Programming**

## **1. Concept of 0/1 Knapsack**

**0/1 Knapsack is an optimization problem in which we select items to maximize total profit without exceeding the given knapsack capacity.**

The important condition is:

- **0:** Do not select the item.
- **1:** Select the complete item.

Unlike Fractional Knapsack, we cannot divide an item into smaller parts.

### **2. Understand with an example**

Suppose we have a bag with a maximum capacity of 50 kg and three items.

| Item | Weight | Profit |
| ---- | ------ | ------ |
| A    | 10 kg  | ₹60    |
| B    | 20 kg  | ₹100   |
| C    | 30 kg  | ₹120   |

Our objective is to select items to get maximum profit without exceeding 50 kg.

### **3. Dynamic Programming Approach**

Dynamic Programming (DP) solves a problem by breaking it into smaller subproblems and storing their results to avoid repeated calculations.

For every item, we have two choices:

1. **Exclude the item:** Keep the previous maximum profit.
2. **Include the item:** Add its profit to the best solution for the remaining capacity.

We select whichever gives maximum profit.

### **4. Mathematical Formula**

Let:

- \\(n\\) = Number of items.
- \\(W\\) = Knapsack capacity.
- \\(wt[i]\\) = Weight of item.
- \\(p[i]\\) = Profit of item.
- \\(DP[i][w]\\) = Maximum profit using the first \\(i\\) items with capacity \\(w\\).

The recurrence relation is:

\\[ DP[i][w]= \begin{cases} 0 & i=0 \text{ or } w=0\\\ DP[i-1][w] & wt[i]>w\\\ \max(DP[i-1][w],p[i]+DP[i-1][w-wt[i]]) & wt[i]\leq w \end{cases} \\]

**Remember:** We use the maximum of including and excluding an item.



## **7. Step-by-step explanation**

For the above example:

| Item | Weight | Profit |
| ---- | ------ | ------ |
| A    | 10     | 60     |
| B    | 20     | 100    |
| C    | 30     | 120    |

Knapsack capacity = 50 kg.

The best selection is:

- Select item B = 20 kg, profit ₹100.
- Select item C = 30 kg, profit ₹120.
- Total weight = 50 kg.
- Total profit = ₹220.

Maximum Profit = ₹220

## **8. Complexity Analysis**

Let \\(n\\) be the number of items and \\(W\\) be the knapsack capacity.

| Complexity | Value               | Explanation                         |
| ---------- | ------------------- | ----------------------------------- |
| Time       | \\(O(n \times W)\\) | Two nested loops fill the DP table. |
| Space      | \\(O(n \times W)\\) | A 2D DP table is created.           |

## **9. Difference Between Fractional and 0/1 Knapsack**

| Fractional Knapsack        | 0/1 Knapsack                     |
| -------------------------- | -------------------------------- |
| Greedy method              | Dynamic Programming              |
| Items can be divided.      | Items cannot be divided.         |
| Partial selection allowed. | Only complete selection allowed. |
| Time: \\(O(n\log n)\\)     | Time: \\(O(nW)\\)                |

## **10. Conclusion**

The 0/1 Knapsack problem is solved using Dynamic Programming. Each item is either selected completely or rejected. The DP table stores previously calculated results and avoids repeated calculations. The algorithm finds the maximum possible profit within the given capacity.
